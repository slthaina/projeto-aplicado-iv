# Resumo

Este projeto investiga mudanças na dinâmica temporal de dengue, zika e chikungunya em cinco municípios brasileiros e sua associação com variáveis climáticas. Foram utilizados dados semanais do InfoDengue, organizados para Brasília, Manaus, Porto Alegre, Recife e São Paulo, no período de 2010 a 2024. O processamento incluiu a integração das bases epidemiológicas e climáticas, a avaliação da qualidade dos dados, a análise de padrões sazonais, o cálculo de indicadores de temporada, a detecção de pontos de mudança e a investigação exploratória de associações entre variáveis climáticas e casos em diferentes defasagens temporais. A sazonalidade foi examinada em janelas de 2010–2014, 2015–2019 e 2020–2024. A detecção de mudanças foi realizada sobre séries dessazonalizadas por STL com o método PELT, e as associações climáticas foram exploradas por correlações de Pearson em defasagens de até oito semanas. Os resultados foram organizados em tabelas e gráficos. A auditoria também identificou limitações de qualidade em algumas variáveis climáticas, que exigem cautela na interpretação. O projeto tem caráter observacional, portanto as associações identificadas não demonstram causalidade. # Referencial Teórico

Dengue, zika e chikungunya compartilham o vetor *Aedes aegypti*, cuja biologia e capacidade de transmissão dependem das condições ambientais. Modelos mecanísticos indicam que a adequação térmica à transmissão é não linear e difere entre os vírus (MORDECAI et al., 2017; 2019). Projeções globais também sugerem que as mudanças climáticas podem redistribuir o risco de transmissão de arboviroses (RYAN et al., 2019). Este projeto investiga se padrões temporais compatíveis com essas relações podem ser observados em séries de vigilância, sem presumir causalidade.

O InfoDengue integra informações epidemiológicas e ambientais e utiliza *nowcasting* para lidar com atrasos de notificação (CODEÇO et al., 2016; 2018). Como as estimativas podem ser revisadas retrospectivamente, os casos estimados são tratados como medida dinâmica. Bastos et al. (2019) discutem abordagens de modelagem para correção de atrasos em dados de vigilância.

Estudos anteriores exploraram a relação entre clima e arboviroses por diferentes métodos. Cazelles et al. (2005) utilizaram análise de *wavelets* para examinar relações não estacionárias entre clima e dengue. Lowe et al. (2021) modelaram efeitos combinados de riscos hidrometeorológicos e urbanização sobre o risco de dengue no Brasil. Modelos de defasagem distribuída não linear permitem representar relações não lineares entre exposição e resposta ao longo do tempo (GASPARRINI; ARMSTRONG; KENWARD, 2010). Neste projeto, a associação é inicialmente explorada por correlações em diferentes defasagens, método mais simples que não controla, por si só, fatores de confusão nem demonstra causalidade.

A decomposição STL separa tendência, sazonalidade e resíduo, facilitando a análise de padrões temporais (CLEVELAND et al., 1990). Para detectar possíveis alterações estruturais, o método PELT oferece uma abordagem computacionalmente eficiente, embora os resultados dependam da função de custo e da penalização utilizadas (KILLICK; FEARNHEAD; ECKLEY, 2012; TRUONG; OUDRE; VAYATIS, 2020).

O planejamento inicial prevê modelagem preditiva com SARIMAX e XGBoost. O SARIMAX permite incorporar variáveis exógenas, enquanto o XGBoost representa relações não lineares e interações entre atributos (HYNDMAN; ATHANASOPOULOS, 2021; CHEN; GUESTRIN, 2016). A comparação entre esses modelos será apresentada como resultado apenas quando houver execução e validação documentadas.

## Preparação e auditoria

O fluxo inclui padronização dos dados, identificação do município a partir do nome dos arquivos, integração das doenças e associação com as variáveis climáticas. A base climática foi obtida a partir dos arquivos de dengue como fonte comum, pois os dados climáticos associados às séries de zika apresentavam congelamento em data anterior. Essa decisão evita combinar séries climáticas inconsistentes entre doenças, mas exige cautela na interpretação.

A auditoria identificou comportamentos potencialmente artificiais ou interpolados em `tempmin` em semanas nas quais outras variáveis climáticas estavam ausentes. Também foi identificado um valor fisicamente suspeito em Porto Alegre, na semana epidemiológica 201352, com temperatura mínima, média e máxima iguais a 38 °C. Esses registros são tratados como limitações da base, e não interpretados automaticamente como condições climáticas reais.

O INMET/BDMEP foi considerado como fonte complementar, mas não foi integrado nesta etapa devido à limitação de tempo, à inconsistência entre formatos de arquivos históricos e à descontinuidade dos dados de estações. A variável `casos_est` pode sofrer revisões retrospectivas devido ao *nowcasting*.

## Limitações dos dados

As principais limitações são a subnotificação; a revisão retrospectiva das estimativas de casos; possíveis mudanças na vigilância e na cobertura de testagem; diferenças de completude entre municípios e doenças; ausência de variáveis socioambientais, como saneamento, urbanização e controle vetorial; e incertezas em parte das variáveis climáticas. # Modelos

## Indicadores de sazonalidade

A sazonalidade foi examinada em três janelas: 2010–2014, 2015–2019 e 2020–2024. Foram calculados indicadores de temporada por município, doença e ano, utilizando um limiar baseado no percentil 75. Os indicadores incluem início, pico, fim, duração e intensidade. O limiar pode apresentar limitações em séries com baixa incidência ou muitos zeros, especialmente para zika e chikungunya, e deve ser interpretado considerando a qualidade de cada série.

## Decomposição e pontos de mudança

A decomposição STL foi utilizada para separar tendência, sazonalidade e resíduo. Em seguida, o método PELT foi aplicado à série dessazonalizada para identificar possíveis pontos de mudança. Esses pontos representam alterações estatísticas na série e não demonstram, isoladamente, que o clima tenha causado a mudança.

## Associação com variáveis climáticas

Foram construídas defasagens de 0 a 8 semanas para variáveis climáticas disponíveis. A relação entre clima e casos foi explorada por meio de correlações de Pearson. A análise descreve associações lineares em diferentes defasagens, mas não controla todos os fatores de confusão e não deve ser interpretada como medida direta de um atraso biológico do vetor.

## Modelo base aplicado

O método estatístico aplicado nesta implementação parcial foi o PELT (*Pruned Exact Linear Time*), utilizado para identificar possíveis pontos de mudança nas séries temporais. A análise foi realizada após a decomposição STL, com o objetivo de reduzir a influência da sazonalidade anual esperada. Os pontos detectados são resultados exploratórios e dependem dos parâmetros adotados e da qualidade das séries.

## Modelagem preditiva prevista

## `src/analisar_lags_climaticos.py`


````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

def analisar_lags():
    print("Carregando base de dados integrada...")
    df = pd.read_csv("data/processed/epidemio_clima_semanal.csv")
    
    # Ordenar chronologicamente por município e SE
    df = df.sort_values(["municipio_geocodigo", "SE"]).reset_index(drop=True)
    
    # Criar lags de 0 a 8 semanas para variáveis climáticas principais
    max_lags = 8
    variaveis_clima = ['tempmed', 'tempmin', 'umidmed']
    doencas = ['dengue', 'zika', 'chikungunya']
    
    print("Gerando colunas de defasagem (lags) para temperatura e humidade...")
    for var in variaveis_clima:
        for lag in range(1, max_lags + 1):
            df[f'{var}_lag_{lag}'] = df.groupby('municipio_geocodigo')[var].shift(lag)
            
    # Salvar a base atualizada com os lags para a etapa preditiva futura
    os.makedirs("data/processed", exist_ok=True)
    df.to_csv("data/processed/epidemio_clima_lags.csv", index=False)
    print("Base com lags salva em: data/processed/epidemio_clima_lags.csv")
    
    # Calcular a correlação cruzada entre casos e lags climáticos
    resultados_corr = []
    
    for doenca in doencas:
        col_casos = f'casos_est_{doenca}'
        if col_casos not in df.columns:
            continue
            
        for var in variaveis_clima:
            # Lag 0 (mesma semana)
            corr_lag0 = df[col_casos].corr(df[var])
            resultados_corr.append({
                'doenca': doenca,
                'variavel_clima': var,
                'lag_semanas': 0,
                'correlacao': corr_lag0
            })
            
            # Lags 1 a 8
            for lag in range(1, max_lags + 1):
                col_lag = f'{var}_lag_{lag}'
                corr = df[col_casos].corr(df[col_lag])
                resultados_corr.append({
                    'doenca': doenca,
                    'variavel_clima': var,
                    'lag_semanas': lag,
                    'correlacao': corr
                })
                
    df_corr = pd.DataFrame(resultados_corr)
    df_corr.to_csv("data/processed/correlacao_lags_clima.csv", index=False)
    print("Matriz de correlação salva em: data/processed/correlacao_lags_clima.csv")
    
    # Plotar o gráfico de lags para Temperatura Média
    print("Gerando gráfico de correlação por defasagem...")
    fig, ax = plt.subplots(figsize=(10, 6))
    
    df_tempmed = df_corr[df_corr['variavel_clima'] == 'tempmed']
    
    for doenca in doencas:
        dados_d = df_tempmed[df_tempmed['doenca'] == doenca]
        if not dados_d.empty:
            ax.plot(dados_d['lag_semanas'], dados_d['correlacao'], marker='o', linewidth=2, label=doenca.capitalize())
            
    ax.set_title('Correlação Cruzada: Temperatura Média vs Casos Estimados (por Lag)', fontsize=14, fontweight='bold', pad=15)
    ax.set_xlabel('Defasagem / Lag em Semanas (0 = Mesma semana)', fontweight='bold')
    ax.set_ylabel('Coeficiente de Correlação de Pearson', fontweight='bold')
    ax.set_xticks(range(0, max_lags + 1))
    ax.grid(True, linestyle='--', alpha=0.5)
    ax.legend(loc='upper right')
    
    caminho_grafico = "data/processed/grafico_lags_clima.png"
    plt.savefig(caminho_grafico, dpi=300, bbox_inches='tight')
    plt.close()
    print(f"Gráfico de lags salvo em: {caminho_grafico}")

if __name__ == "__main__":
    analisar_lags()
````

## `src/analise_sazonalidade.py`


````python
import pandas as pd
import matplotlib.pyplot as plt

def plotar_sazonalidade_periodos():
    print("Carregando base de dados integrada...")
    df = pd.read_csv("data/processed/epidemio_clima_semanal.csv")

    # Extrair o ano e a semana do ano a partir do código SE (ex: 202045 -> Ano 2020, Semana 45)
    df['ano'] = df['SE'].astype(str).str[:4].astype(int)
    df['semana_do_ano'] = df['SE'].astype(str).str[-2:].astype(int)

    # Função para classificar as linhas nos três períodos de interesse
    def classificar_periodo(ano):
        if 2010 <= ano <= 2014: return '2010-2014'
        elif 2015 <= ano <= 2019: return '2015-2019'
        elif 2020 <= ano <= 2024: return '2020-2024'
        return None

    df['periodo'] = df['ano'].apply(classificar_periodo)
    df = df.dropna(subset=['periodo']) # Remove anos fora do escopo

    periodos = ['2010-2014', '2015-2019', '2020-2024']
    
    # Criar uma figura com 3 gráficos empilhados (3 linhas, 1 coluna)
    print("Gerando os gráficos segmentados...")
    fig, axes = plt.subplots(3, 1, figsize=(12, 16), sharex=True)
    cols_agrupar = ['casos_est_dengue', 'casos_est_zika', 'casos_est_chikungunya', 'tempmed']

    for i, p in enumerate(periodos):
        df_p = df[df['periodo'] == p]
        sazonalidade = df_p.groupby('semana_do_ano')[cols_agrupar].mean().reset_index()
        
        ax1 = axes[i]
        
        # Eixo esquerdo: Casos
        ax1.plot(sazonalidade['semana_do_ano'], sazonalidade['casos_est_dengue'], label='Dengue', color='#d62728', linewidth=2)
        ax1.plot(sazonalidade['semana_do_ano'], sazonalidade['casos_est_zika'], label='Zika', color='#1f77b4', linewidth=2)
        ax1.plot(sazonalidade['semana_do_ano'], sazonalidade['casos_est_chikungunya'], label='Chikungunya', color='#ff7f0e', linewidth=2)
        ax1.set_ylabel(f'Média de Casos\n({p})', fontweight='bold')
        ax1.grid(True, linestyle='--', alpha=0.5)
        
        # Posicionar legenda de casos apenas no primeiro gráfico para evitar poluição visual
        if i == 0:
            ax1.legend(loc='upper left')
        
        # Eixo direito: Temperatura
        ax2 = ax1.twinx()
        ax2.plot(sazonalidade['semana_do_ano'], sazonalidade['tempmed'], label='Temp. Média (°C)', color='#7f7f7f', linestyle='dashed', linewidth=2)
        ax2.set_ylabel('Temp. Média (°C)', fontweight='bold', color='#7f7f7f')
        ax2.tick_params(axis='y', labelcolor='#7f7f7f')
        
        if i == 0:
            ax2.legend(loc='upper right')

    # Ajustes finais no layout
    axes[-1].set_xlabel('Semana Epidemiológica do Ano (1 a 53)', fontweight='bold')
    plt.suptitle('Sazonalidade: Comportamento Médio de Arboviroses vs Temperatura', fontsize=16, y=0.92, fontweight='bold')
    
    # Salvar a imagem
    caminho_saida = "data/processed/grafico_sazonalidade_periodos.png"
    plt.savefig(caminho_saida, dpi=300, bbox_inches='tight')
    print(f"Sucesso! Gráfico salvo em: {caminho_saida}")

if __name__ == "__main__":
    plotar_sazonalidade_periodos()
````

## `src/calcular_indicadores.py`


````python
import pandas as pd
import numpy as np
import os

def calcular_indicadores_temporada():
    print("Carregando base de dados integrada...")
    df = pd.read_csv("data/processed/epidemio_clima_semanal.csv")

    # Extrair Ano e Semana Epidemiológica da coluna SE
    df['ano'] = df['SE'].astype(str).str[:4].astype(int)
    df['semana_do_ano'] = df['SE'].astype(str).str[-2:].astype(int)

    # Função de classificação de períodos
    def classificar_periodo(ano):
        if 2010 <= ano <= 2014: return '2010-2014'
        elif 2015 <= ano <= 2019: return '2015-2019'
        elif 2020 <= ano <= 2024: return '2020-2024'
        return None

    df['periodo'] = df['ano'].apply(classificar_periodo)
    df = df.dropna(subset=['periodo'])

    doencas = ['dengue', 'zika', 'chikungunya']
    municipios = df['municipio_geocodigo'].unique()
    resultados = []

    for muni in municipios:
        df_muni = df[df['municipio_geocodigo'] == muni]
        
        for doenca in doencas:
            col_casos = f'casos_est_{doenca}'
            if col_casos not in df_muni.columns:
                continue

            # 1. Limiar: Percentil 75 histórico do município para a doença
            limiar_p75 = df_muni[col_casos].quantile(0.75)

            for (periodo, ano), grupo in df_muni.groupby(['periodo', 'ano']):
                grupo = grupo.sort_values('semana_do_ano')
                casos = grupo[col_casos].values
                semanas = grupo['semana_do_ano'].values

                if len(casos) == 0 or np.isnan(casos).all() or np.sum(casos) == 0:
                    continue

                # Indicador 2: Pico
                idx_pico = np.argmax(casos)
                semana_pico = semanas[idx_pico]
                valor_pico = casos[idx_pico]

                # Condição de estar acima do limiar
                acima_limiar = casos > limiar_p75

                # Indicador 1: Início (primeira semana com 2+ semanas consecutivas acima do P75)
                semana_inicio = np.nan
                for k in range(len(acima_limiar) - 1):
                    if acima_limiar[k] and acima_limiar[k + 1]:
                        semana_inicio = semanas[k]
                        break

                # Indicador 3: Fim (última semana acima do P75)
                indices_acima = np.where(acima_limiar)[0]
                semana_fim = semanas[indices_acima[-1]] if len(indices_acima) > 0 else np.nan

                # Indicador 4: Duração em semanas
                duracao = (semana_fim - semana_inicio + 1) if (not np.isnan(semana_inicio) and not np.isnan(semana_fim) and semana_fim >= semana_inicio) else np.nan

                # Indicador 5: Intensidade (Casos Acumulados no Ano)
                intensidade_acumulada = np.sum(casos)

                resultados.append({
                    'municipio_geocodigo': muni,
                    'doenca': doenca,
                    'periodo': periodo,
                    'ano': ano,
                    'limiar_p75_historico': round(limiar_p75, 2),
                    'semana_inicio': semana_inicio,
                    'semana_pico': semana_pico,
                    'valor_pico': round(valor_pico, 1),
                    'semana_fim': semana_fim,
                    'duracao_semanas': duracao,
                    'intensidade_acumulada': round(intensidade_acumulada, 1)
                })

    df_res = pd.DataFrame(resultados)
    
    # Salvar resultados detalhados por ano
    caminho_saida = "data/processed/indicadores_sazonalidade.csv"
    os.makedirs("data/processed", exist_ok=True)
    df_res.to_csv(caminho_saida, index=False)
    print(f"Indicadores anuais salvos em: {caminho_saida}")

    # Resumo consolidado por Período (Média dos indicadores)
    resumo_periodos = df_res.groupby(['doenca', 'periodo'])[[
        'semana_inicio', 'semana_pico', 'duracao_semanas', 'intensidade_acumulada'
    ]].mean().round(1)
    
    print("\n=== RESUMO MÉDIO DOS INDICADORES POR PERÍODO ===")
    print(resumo_periodos.to_string())

if __name__ == "__main__":
    calcular_indicadores_temporada()
````

## `src/coletar_epidemio.py`


````python
"""
Coleta epidemiológica (InfoDengue) por município x doença x período.

- Um arquivo CSV por combinação, salvo exatamente como a API devolveu.
- Se o arquivo já existe, é pulado (a coleta pode ser retomada e nunca sobrescreve).
- Cada tentativa é registrada em um log (data/hora, parâmetros, status, nº de linhas).
- "Sem linhas" (a API não devolveu semanas) fica registrado separadamente de
  "linhas com casos = 0" (a API devolveu a semana com zero casos).
"""
import csv
import io
import sys
import time
from datetime import datetime
from pathlib import Path

import requests

sys.path.append(str(Path(__file__).resolve().parent.parent))
import config as cfg


def baixar(geocode, doenca, ano_ini, ano_fim):
    params = {
        "geocode": geocode,
        "disease": doenca,
        "format": "csv",
        "ew_start": 1,
        "ew_end": 53,
        "ey_start": ano_ini,
        "ey_end": ano_fim,
    }
    ultimo_erro = ""
    for tentativa in range(1, cfg.TENTATIVAS + 1):
        try:
            r = requests.get(cfg.URL_API, params=params, timeout=cfg.TIMEOUT)
            if r.status_code == 200:
                return r.text, "ok", ""
            ultimo_erro = f"HTTP {r.status_code}"
        except requests.RequestException as e:
            ultimo_erro = str(e)[:200]
        time.sleep(cfg.PAUSA_SEGUNDOS * tentativa * 2)
    return None, "falha", ultimo_erro


def contar_linhas(texto):
    leitor = csv.reader(io.StringIO(texto))
    return max(sum(1 for _ in leitor) - 1, 0)  # desconta o cabeçalho


def main():
    cfg.DIR_LOGS.mkdir(parents=True, exist_ok=True)
    caminho_log = cfg.DIR_LOGS / "log_coleta_epidemio.csv"
    novo_log = not caminho_log.exists()

    combinacoes = [
        (g, nome, d, p, a, b)
        for g, nome in cfg.MUNICIPIOS.items()
        for d in cfg.DOENCAS
        for p, (a, b) in cfg.PERIODOS.items()
    ]
    print(f"{len(combinacoes)} combinações a verificar.")

    with open(caminho_log, "a", newline="", encoding="utf-8") as f_log:
        log = csv.writer(f_log)
        if novo_log:
            log.writerow(["data_hora", "geocode", "municipio", "doenca",
                          "periodo", "status", "n_linhas", "detalhe"])

        for i, (geocode, nome, doenca, periodo, a, b) in enumerate(combinacoes, 1):
            pasta = cfg.DIR_RAW / doenca
            pasta.mkdir(parents=True, exist_ok=True)
            destino = pasta / f"{geocode}_{nome}_{doenca}_{periodo}.csv"
            rotulo = f"[{i}/{len(combinacoes)}] {nome} | {doenca} | {periodo}"

            if destino.exists():
                print(f"{rotulo}: já existe, pulando")
                continue

            texto, status, detalhe = baixar(geocode, doenca, a, b)
            agora = datetime.now().isoformat(timespec="seconds")

            if status == "ok":
                n = contar_linhas(texto)
                destino.write_text(texto, encoding="utf-8")
                if n == 0:
                    status, detalhe = "sem_linhas", "API respondeu sem semanas"
                print(f"{rotulo}: {status} ({n} linhas)")
                log.writerow([agora, geocode, nome, doenca, periodo, status, n, detalhe])
            else:
                print(f"{rotulo}: FALHA ({detalhe})")
                log.writerow([agora, geocode, nome, doenca, periodo, status, "", detalhe])

            f_log.flush()
            time.sleep(cfg.PAUSA_SEGUNDOS)

    print("Fim. Rode de novo para tentar apenas o que falhou.")


if __name__ == "__main__":
    main()
````

## `src/detectar_changepoint.py`


````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import STL
import ruptures as rpt
import os

def detectar_changepoint_pelt():
    print("Carregando base de dados integrada...")
    df = pd.read_csv("data/processed/epidemio_clima_semanal.csv")
    
    # Ordenar por Semana Epidemiológica
    df = df.sort_values("SE").reset_index(drop=True)
    
    # Agrupar a soma dos casos por semana para ter a série global dos municípios
    doencas = ['dengue', 'zika', 'chikungunya']
    
    os.makedirs("data/processed", exist_ok=True)
    
    for doenca in doencas:
        col_casos = f'casos_est_{doenca}'
        if col_casos not in df.columns:
            continue
            
        # Criar série agregada no tempo
        serie_temporal = df.groupby('SE')[col_casos].sum().reset_index()
        
        # Preencher eventuais NaNs com 0 para a decomposição
        y = serie_temporal[col_casos].fillna(0).values
        
        if len(y) < 104 or np.sum(y) == 0:  # Necessário pelo menos 2 anos de dados
            print(f"Série de {doenca} insuficiente para decomposição STL/PELT.")
            continue
            
        print(f"\n--- Processando {doenca.upper()} ---")
        
        # 1. Decomposição STL (período de 52 semanas para capturar o ciclo anual)
        stl = STL(y, period=52, robust=True)
        res = stl.fit()
        
        # Serie dessazonalizada = Tendência + Resíduo (ou apenas Tendência)
        tendencia = res.trend
        
        # 2. Algoritmo PELT (modelo 'rbf' para mudanças em média/variância)
        # pen (penalidade): controla a sensibilidade a novas quebras
        algo = rpt.Pelt(model="rbf", min_size=26).fit(tendencia)
        pontos_mudanca = algo.predict(pen=10)
        
        # Mapear os índices dos pontos para as Semanas Epidemiológicas (SE)
        semanas_mudanca = [serie_temporal['SE'].iloc[idx-1] for idx in pontos_mudanca if idx < len(serie_temporal)]
        
        print(f"Pontos de mudança detectados (SE): {semanas_mudanca}")
        
        # 3. Plotar resultado
        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8), sharex=True)
        
        # Gráfico 1: Série Original de Casos
        ax1.plot(serie_temporal['SE'].astype(str), y, label=f'Casos Estimados ({doenca})', color='black', alpha=0.6)
        ax1.set_ylabel('Nº de Casos', fontweight='bold')
        ax1.set_title(f'Detecção de Changepoint (PELT) - {doenca.capitalize()}', fontsize=14, fontweight='bold')
        ax1.grid(True, linestyle='--', alpha=0.5)
        ax1.legend(loc='upper left')
        
        # Gráfico 2: Componente de Tendência Dessazonalizada + Pontos de Ruptura
        ax2.plot(serie_temporal['SE'].astype(str), tendencia, label='Tendência (Dessazonalizada - STL)', color='blue', linewidth=2)
        
        # Desenhar linhas verticais para cada ponto de ruptura encontrado
        for idx in pontos_mudanca[:-1]:  # Ignora o último elemento que é o tamanho total da série
            se_rotulo = serie_temporal['SE'].iloc[idx-1]
            ax1.axvline(x=idx-1, color='red', linestyle='--', linewidth=1.5)
            ax2.axvline(x=idx-1, color='red', linestyle='--', linewidth=1.5, label='Changepoint' if idx == pontos_mudanca[0] else "")
            
        ax2.set_ylabel('Tendência Dessazonalizada', fontweight='bold')
        ax2.set_xlabel('Semana Epidemiológica (SE)', fontweight='bold')
        ax2.grid(True, linestyle='--', alpha=0.5)
        ax2.legend(loc='upper left')
        
        # Reduzir quantidade de rótulos no eixo X para legibilidade
        passo_ticks = max(1, len(serie_temporal) // 10)
        plt.xticks(ticks=range(0, len(serie_temporal), passo_ticks), 
                   labels=serie_temporal['SE'].astype(str).iloc[::passo_ticks], 
                   rotation=45)
        
        caminho_grafico = f"data/processed/changepoint_{doenca}.png"
        plt.savefig(caminho_grafico, dpi=300, bbox_inches='tight')
        plt.close()
        print(f"Gráfico salvo em: {caminho_grafico}")

if __name__ == "__main__":
    detectar_changepoint_pelt()
````

## `src/integrar_bases.py`


````python
import pandas as pd
import glob
import os

def integrar_dados():
    # Listar arquivos epidemiológicos
    path_dengue = "data/raw/epidemio/dengue/*.csv"
    
    # 1. Carregar base de Dengue (Casos + Clima Oficial)
    dfs_d = []
    for f in glob.glob(path_dengue):
        df = pd.read_csv(f)
        # Extrai o geocódigo do nome do arquivo (ex: "2611606_Recife..." -> "2611606")
        geocode = os.path.basename(f).split('_')[0]
        df['municipio_geocodigo'] = geocode
        dfs_d.append(df)
        
    if not dfs_d:
        print("Erro: Nenhum arquivo encontrado em data/raw/epidemio/dengue/")
        return
        
    dengue = pd.concat(dfs_d, ignore_index=True)
    
    # Colunas climáticas do InfoDengue
    cols_clima = ["SE", "municipio_geocodigo", "tempmin", "tempmed", "tempmax", "umidmin", "umidmed", "umidmax"]
    
    # Colunas base e métricas de casos
    cols_base_casos = ["SE", "municipio_geocodigo"]
    cols_metricas = ["casos_est", "casos", "notif_ev"]
    
    # Garante que só vai tentar puxar as colunas que realmente existem no CSV
    cols_dengue_casos = cols_base_casos + [c for c in cols_metricas if c in dengue.columns]
    
    # Isolar Clima da Dengue
    clima = dengue[cols_clima].drop_duplicates(subset=["SE", "municipio_geocodigo"])
    
    # Extrair Casos de Dengue e renomear colunas
    renames_d = {c: f"{c}_dengue" for c in cols_metricas}
    dengue_casos = dengue[cols_dengue_casos].rename(columns=renames_d)
    
    # 2. Funcoes auxiliares para Zika e Chikungunya
    def carregar_casos_outras(doenca):
        files = glob.glob(f"data/raw/epidemio/{doenca}/*.csv")
        if not files:
            return pd.DataFrame()
        dfs = []
        for f in files:
            df = pd.read_csv(f)
            df['municipio_geocodigo'] = os.path.basename(f).split('_')[0]
            dfs.append(df)
        df_all = pd.concat(dfs, ignore_index=True)
        
        # Filtra apenas as colunas que existem
        cols_atuais = cols_base_casos + [c for c in cols_metricas if c in df_all.columns]
        renames = {c: f"{c}_{doenca}" for c in cols_metricas}
        return df_all[cols_atuais].rename(columns=renames)
    
    zika_casos = carregar_casos_outras("zika")
    chik_casos = carregar_casos_outras("chikungunya")
    
    # 3. Merge Final
    # Junta Dengue + Clima
    base = pd.merge(dengue_casos, clima, on=["SE", "municipio_geocodigo"], how="outer")
    
    # Junta Zika e Chikungunya se existirem
    if not zika_casos.empty:
        base = pd.merge(base, zika_casos, on=["SE", "municipio_geocodigo"], how="left")
    if not chik_casos.empty:
        base = pd.merge(base, chik_casos, on=["SE", "municipio_geocodigo"], how="left")
        
    os.makedirs("data/processed", exist_ok=True)
    caminho_saida = "data/processed/epidemio_clima_semanal.csv"
    base.to_csv(caminho_saida, index=False)
    
    print(f"Sucesso! Base integrada com {len(base)} linhas salva em: {caminho_saida}")

if __name__ == "__main__":
    integrar_dados()
````

## `src/validar_epidemio.py`


````python
"""
Validação e consolidação dos dados epidemiológicos brutos.

Gera:
- data/processed/epidemio_semanal.csv : tabela única (município x doença x semana).
  Semanas ausentes entram como linhas com casos vazios (NaN) e semana_ausente=True.
  Elas NUNCA são preenchidas com zero.
- outputs/relatorio_validacao.csv : um resumo por município x doença x período.

Os arquivos em data/raw não são alterados.
"""
import sys
from pathlib import Path

import pandas as pd

sys.path.append(str(Path(__file__).resolve().parent.parent))
import config as cfg

# Regras de classificação (documentar no projeto)
LIMIAR_SEM_CIRCULACAO = 0.95  # proporção de semanas com 0 casos
LIMIAR_ESPARSA = 0.50

COLUNAS_UTEIS = ["data_iniSE", "SE", "casos", "casos_est", "p_inc100k", "pop"]


def ler_arquivo(geocode, nome, doenca, periodo):
    caminho = cfg.DIR_RAW / doenca / f"{geocode}_{nome}_{doenca}_{periodo}.csv"
    if not caminho.exists():
        return None
    d = pd.read_csv(caminho)
    colunas = [c for c in COLUNAS_UTEIS if c in d.columns]
    d = d[colunas].copy()
    d["geocode"], d["municipio"] = geocode, nome
    d["doenca"], d["periodo"] = doenca, periodo
    return d


def main():
    pedacos = []
    for geocode, nome in cfg.MUNICIPIOS.items():
        for doenca in cfg.DOENCAS:
            for periodo in cfg.PERIODOS:
                d = ler_arquivo(geocode, nome, doenca, periodo)
                if d is None:
                    print(f"AVISO: arquivo ausente: {nome} | {doenca} | {periodo}")
                    continue
                pedacos.append(d)

    bruto = pd.concat(pedacos, ignore_index=True)

    # Semanas esperadas por período = todas as SE que aparecem em qualquer série do período
    esperadas = bruto.groupby("periodo")["SE"].apply(lambda s: sorted(s.unique())).to_dict()

    tabelas, resumo = [], []
    for (geocode, nome, doenca, periodo), d in bruto.groupby(
            ["geocode", "municipio", "doenca", "periodo"]):
        dup = int(d.duplicated("SE").sum())
        d = d.drop_duplicates("SE")
        grade = pd.DataFrame({"SE": esperadas[periodo]})
        full = grade.merge(d, on="SE", how="left")
        full["semana_ausente"] = full["casos"].isna()
        for col, val in [("geocode", geocode), ("municipio", nome),
                         ("doenca", doenca), ("periodo", periodo)]:
            full[col] = val

        n_esp = len(full)
        n_aus = int(full["semana_ausente"].sum())
        obs = full.loc[~full["semana_ausente"], "casos"]
        prop_zeros = float((obs == 0).mean()) if len(obs) else float("nan")
        # série truncada: ausências contíguas no final
        truncada = bool(full["semana_ausente"].iloc[-1]) if n_esp else False

        if prop_zeros >= LIMIAR_SEM_CIRCULACAO:
            classe = "sem_circulacao_relevante"
        elif prop_zeros >= LIMIAR_ESPARSA:
            classe = "esparsa"
        else:
            classe = "ok"

        igual = (bool((obs == full.loc[~full["semana_ausente"], "casos_est"]).all())
                 if "casos_est" in full else None)

        resumo.append({
            "municipio": nome, "doenca": doenca, "periodo": periodo,
            "semanas_esperadas": n_esp, "semanas_ausentes": n_aus,
            "serie_truncada": truncada, "duplicatas_removidas": dup,
            "soma_casos": float(obs.sum()), "semanas_zero": int((obs == 0).sum()),
            "prop_zeros": round(prop_zeros, 3), "classificacao": classe,
            "casos_est_igual_casos": igual,
        })
        full["classificacao_serie"] = classe
        tabelas.append(full)

    tabela = pd.concat(tabelas, ignore_index=True)
    tabela["ano_se"] = tabela["SE"] // 100
    tabela["semana_se"] = tabela["SE"] % 100

    cfg.DIR_RAW.parent.parent.joinpath("processed").mkdir(parents=True, exist_ok=True)
    cfg.DIR_LOGS.parent.mkdir(parents=True, exist_ok=True)
    saida_tab = cfg.RAIZ / "data" / "processed" / "epidemio_semanal.csv"
    saida_rel = cfg.RAIZ / "outputs" / "relatorio_validacao.csv"
    tabela.to_csv(saida_tab, index=False)
    pd.DataFrame(resumo).to_csv(saida_rel, index=False)

    print(f"Tabela: {saida_tab} ({len(tabela)} linhas)")
    print(f"Relatório: {saida_rel}")
    rel = pd.DataFrame(resumo)
    print("\nClassificação das séries:")
    print(rel.groupby(["doenca", "periodo"])["classificacao"].value_counts().to_string())
    print("\nSéries com semanas ausentes:")
    print(rel.loc[rel["semanas_ausentes"] > 0,
                  ["municipio", "doenca", "periodo", "semanas_ausentes"]].to_string(index=False))


if __name__ == "__main__":
    main()
````
